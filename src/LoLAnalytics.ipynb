```python
!pip install scikit-fuzzy
import pandas as pd
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt

# ----------------------------
# 1. Definição da variável de saída
# ----------------------------
uDj = np.arange(0, 1.05, 0.05)
Dj = ctrl.Consequent(uDj, 'desempenho do jogador')
Dj['MR'] = fuzz.trapmf(Dj.universe, [0, 0, 0.1, 0.3])
Dj['R']  = fuzz.trimf(Dj.universe, [0.05, 0.2, 0.4])
Dj['M']  = fuzz.trimf(Dj.universe, [0.3, 0.5, 0.7])
Dj['B']  = fuzz.trimf(Dj.universe, [0.6, 0.7, 0.8])
Dj['MB'] = fuzz.trimf(Dj.universe, [0.7, 0.8, 0.9])
Dj['P']  = fuzz.trapmf(Dj.universe, [0.8, 0.9, 1, 1])
# Opcional: visualize a variável de saída
Dj.view()
plt.title('Conjuntos fuzzy para desempenho do jogador')
plt.show()
```

```python
# ----------------------------
# 2. Carregar dados e gerar variáveis fuzzy de entrada via clusterização
# ----------------------------

# Carregar os dados do CSV
df = pd.read_csv('dados_challenger.csv')

# Lista de variáveis de entrada (ajuste conforme as colunas do seu CSV)
input_vars = ['kills', 'deaths', 'assists', 'cs', 'objective_participation',
              'visionScore', 'goldEarned', 'damageDealtToChampions']

# Agrupa os dados por 'lane' e 'champion_class'
grupos = df.groupby(['lane', 'champion_class'])

# Número de clusters desejados para cada variável (ex: baixo, médio, alto)
n_clusters = 3

# Dicionário para armazenar as variáveis fuzzy geradas para cada grupo e variável
fuzzy_vars = {}

# Itera sobre cada grupo formado por lane e champion_class
for (lane, champion_class), grupo in grupos:
    print(f"Processando grupo: {lane} - {champion_class}")

    for var in input_vars:
        dados = grupo[var].values.astype(np.float64)

        # Verifica se há dados suficientes para a clusterização
        if len(dados) < n_clusters:
            print(f"Dados insuficientes para {var} em {lane} - {champion_class}. Pulando...")
            continue

        # Evita divisão por zero em caso de valores constantes
        if dados.max() - dados.min() == 0:
            print(f"Valores constantes para {var} em {lane} - {champion_class}. Pulando...")
            continue

        # Normaliza os dados para o intervalo [0, 1]
        dados_norm = (dados - dados.min()) / (dados.max() - dados.min())
        dados_norm = np.expand_dims(dados_norm, axis=0)  # Formato necessário para o c-means

        # Aplica o algoritmo fuzzy c-means
        cntr, u, u0, d, jm, p, fpc = fuzz.cluster.cmeans(
            dados_norm, c=n_clusters, m=2, error=0.005, maxiter=1000, init=None)

        # Ordena os centros para definir os limites de forma crescente
        centers = np.sort(cntr.flatten())
        print(f"Centros para {var} em {lane} - {champion_class}: {centers}")

        # Define o universo de discurso para a variável (neste caso, [0, 1])
        universo = np.linspace(0, 1, 100)

        # Cria a variável fuzzy (antecedente) com um nome que identifica o grupo e a variável
        nome_var = f"{lane}_{champion_class}_{var}"
        fuzzy_var = ctrl.Antecedent(universo, nome_var)

        # Define as funções de pertinência para cada cluster:
        # - Primeiro cluster: "baixo" (função trapezoidal)
        # - Cluster intermediário: "médio" (função triangular)
        # - Último cluster: "alto" (função trapezoidal)
        for i, center in enumerate(centers):
            if i == 0:
                mf = fuzz.trapmf(universo, [0, 0, centers[i], centers[i+1]])
                label = f"{var}_baixo"
            elif i == n_clusters - 1:
                mf = fuzz.trapmf(universo, [centers[i-1], centers[i], 1, 1])
                label = f"{var}_alto"
            else:
                mf = fuzz.trimf(universo, [centers[i-1], center, centers[i+1]])
                label = f"{var}_medio"
            fuzzy_var[label] = mf

        # Armazena a variável fuzzy no dicionário para uso posterior
        fuzzy_vars[nome_var] = fuzzy_var

        # Visualize os conjuntos fuzzy gerados para essa variável e grupo
        fuzzy_var.view()
        plt.title(f'Conjuntos fuzzy para {var}: {lane} - {champion_class}')
        plt.show()
```

```python
# ----------------------------
# 3. Geração Automática das Regras de Inferência (Método Adaptado Wang-Mendel)
# ----------------------------

# Dicionário que armazenará as listas de regras para cada grupo (chave: f"{lane}_{champion_class}")
regras_por_grupo = {}

# Itera novamente sobre cada grupo para gerar as regras com base nos dados
for (lane, champion_class), grupo in grupos:
    print(f"\nGerando regras para o grupo: {lane} - {champion_class}")

    # Inicializa a lista de regras para o grupo atual
    regras_por_grupo[f"{lane}_{champion_class}"] = []

    # Calcular estatísticas para normalização de cada input no grupo
    stats = {}
    for var in input_vars:
        valores = grupo[var].values.astype(float)
        stats[var] = {'min': valores.min(), 'max': valores.max()}

    # Dicionário temporário para acumular as condições (combinações dos termos fuzzy) e seus desempenhos
    regras_temp = {}

    # Percorre cada amostra do grupo
    for idx, row in grupo.iterrows():
        condicoes = []    # Armazena o termo fuzzy (rótulo) com maior grau para cada input
        norm_values = {}  # Armazena o valor normalizado para cada input

        # Para cada variável de entrada
        for var in input_vars:
            valor = row[var]
            # Normaliza o valor para [0, 1], tratando valores constantes
            if stats[var]['max'] == stats[var]['min']:
                norm = 0.0  # ou outro valor padrão, como 0.5, conforme o que fizer sentido para o seu domínio
            else:
                norm = (valor - stats[var]['min']) / (stats[var]['max'] - stats[var]['min'])
            norm_values[var] = norm

            # Recupera a variável fuzzy criada para este grupo e variável
            fuzzy_var = fuzzy_vars.get(f"{lane}_{champion_class}_{var}")
            if fuzzy_var is None:
                continue

            # Determina o termo fuzzy com maior grau de pertinência
            melhor_label = None
            melhor_grau = -1
            for label, term in fuzzy_var.terms.items():
                grau = fuzz.interp_membership(fuzzy_var.universe, term.mf, norm)
                if grau > melhor_grau:
                    melhor_grau = grau
                    melhor_label = label
            condicoes.append(melhor_label)

        # Usa a combinação dos termos como chave para a regra
        chave = tuple(condicoes)
        # Define um desempenho hipotético como a média dos inputs normalizados
        desempenho_amostra = np.mean(list(norm_values.values()))

        # Acumula os desempenhos para a mesma combinação de condições
        if chave in regras_temp:
            regras_temp[chave].append(desempenho_amostra)
        else:
            regras_temp[chave] = [desempenho_amostra]

    # Para cada combinação única, determina o desempenho médio e mapeia para um termo de saída
    for condicoes, lista_perf in regras_temp.items():
        desempenho_medio = np.mean(lista_perf)

        # Determina o termo da variável de saída com maior grau para o desempenho_medio
        melhor_label_saida = None
        melhor_grau_saida = -1
        for label, term in Dj.terms.items():
            grau_saida = fuzz.interp_membership(Dj.universe, term.mf, desempenho_medio)
            if grau_saida > melhor_grau_saida:
                melhor_grau_saida = grau_saida
                melhor_label_saida = label

        # Constrói os antecedentes combinando, para cada variável, o termo fuzzy selecionado
        antecedentes = []
        for var, label in zip(input_vars, condicoes):
            fuzzy_var = fuzzy_vars.get(f"{lane}_{champion_class}_{var}")
            if fuzzy_var is not None:
                antecedentes.append(fuzzy_var[label])

        # Combina os antecedentes usando o operador lógico AND
        if antecedentes:
            antecedente_combinado = antecedentes[0]
            for ant in antecedentes[1:]:
                antecedente_combinado = antecedente_combinado & ant

            # Cria a regra fuzzy: se (antecedente_combinado) então (variável de saída com o termo escolhido)
            regra = ctrl.Rule(antecedent=antecedente_combinado, consequent=Dj[melhor_label_saida])
            regras_por_grupo[f"{lane}_{champion_class}"].append(regra)
            print(f"Regra gerada para {lane}-{champion_class}: IF {condicoes} THEN {melhor_label_saida}")

# Exibe o total de regras geradas para cada grupo
for grupo, regras in regras_por_grupo.items():
    print(f"Grupo {grupo} -> Total de regras: {len(regras)}")
```

```python
# ----------------------------
# 4. Criação do sistema de controle fuzzy com as regras geradas
# ----------------------------

# Dicionário para armazenar o simulador do sistema fuzzy para cada grupo
sistemas_por_grupo = {}

# Itera sobre cada grupo (chave: "LANE_CHAMPIONCLASS") e sua lista de regras
for grupo, regras in regras_por_grupo.items():
    # Cria o sistema fuzzy com as regras deste grupo
    sistema_fuzzy_grupo = ctrl.ControlSystem(regras)
    # Cria o simulador do sistema fuzzy para o grupo
    sistema_simulacao_grupo = ctrl.ControlSystemSimulation(sistema_fuzzy_grupo)
    # Armazena no dicionário usando a chave do grupo
    sistemas_por_grupo[grupo] = sistema_simulacao_grupo
    print(f"Sistema fuzzy criado para o grupo: {grupo}")
```

```python
# ----------------------------
# 5. Utilizando dados reais para verificar resultados com normalização
# ----------------------------

# Carrega os dados das partidas do jogador
df_jogador = pd.read_csv('dados_jogador.csv')

# Supondo que as partidas do jogador tenham 'lane' e 'champion_class' definidos,
# usamos os valores da primeira partida para identificar o grupo
lane = df_jogador.iloc[0]['lane']
champion_class = df_jogador.iloc[0]['champion_class']

# Monta a chave do grupo (usando o mesmo padrão definido na criação dos antecedentes)
grupo_key = f"{lane}_{champion_class}"

resultados = []

# Verifica se existe um sistema fuzzy para esse grupo
if grupo_key in sistemas_por_grupo:
    sistema_jogador = sistemas_por_grupo[grupo_key]

    # Processa cada partida
    for idx, partida in df_jogador.iterrows():
        # Dicionário para armazenar os inputs normalizados com os nomes corretos
        inputs_normalizados = {}
        for var in input_vars:
            # Recupera os parâmetros de normalização para o grupo
            min_val = stats[var]['min']
            max_val = stats[var]['max']
            valor = partida[var]
            # Normaliza: se max_val == min_val, define como 0 (ou 0.5, se preferir)
            if max_val == min_val:
                norm = 0.0
            else:
                norm = (valor - min_val) / (max_val - min_val)
            # O nome do input deve coincidir com o formato usado na criação dos antecedentes
            inputs_normalizados[f"{grupo_key}_{var}"] = norm

        # Exibe os inputs normalizados para a partida
        print(f"Partida {idx+1} inputs normalizados:", inputs_normalizados)

        # Atribui os valores normalizados aos inputs do sistema fuzzy
        for input_name, norm_value in inputs_normalizados.items():
            sistema_jogador.input[input_name] = norm_value

        # Computa o resultado fuzzy (a defuzzificação utiliza, por padrão, o centro de gravidade)
        sistema_jogador.compute()

        if 'desempenho do jogador' in sistema_jogador.output:
            desempenho = sistema_jogador.output['desempenho do jogador']
        else:
            # Caso nenhuma regra seja ativada, define um valor padrão (ex.: 0) ou registre uma mensagem
            print(f"Partida {idx+1}: Nenhuma regra ativada. Definindo desempenho padrão = 0.")
            desempenho = 0.0

        resultados.append(desempenho)
        print(f"Partida {idx+1}: Desempenho = {desempenho}\n")

    # Opcional: calcula uma média dos desempenhos
    desempenho_medio = np.mean(resultados)
    print(f"\nDesempenho médio do jogador: {desempenho_medio}")
else:
    print(f"Nenhum sistema fuzzy encontrado para o grupo: {grupo_key}")
```

```python
# ----------------------------
# 6. Geração da Imagem da Tabela
# ----------------------------

# Cria uma cópia do DataFrame original e adiciona a coluna "Desempenho"
df_visual = df_jogador.copy()
df_visual['Desempenho'] = resultados

# Arredonda os valores numéricos para melhor visualização
for col in df_visual.select_dtypes(include=[np.number]).columns:
    df_visual[col] = df_visual[col].round(2)

fig, ax = plt.subplots(figsize=(12, 6))
ax.axis('tight')
ax.axis('off')

# Cria a tabela com os dados do DataFrame visual (dados reais + desempenho)
tabela = ax.table(cellText=df_visual.values,
                  colLabels=df_visual.columns,
                  cellLoc='center',
                  loc='center')

# Ajusta o estilo da tabela
tabela.auto_set_font_size(False)
tabela.set_fontsize(10)
tabela.auto_set_column_width(list(range(len(df_visual.columns))))

# Salva a tabela como imagem
plt.savefig("tabela_desempenho_jogador.png", dpi=300, bbox_inches='tight')
plt.show()
```
